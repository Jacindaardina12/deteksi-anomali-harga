```python
import pandas as pd

# Membaca file Excel
file_path = "data skripsi.xlsx"
df = pd.read_excel(file_path)

df
```

```python
# memilih variabel yang digunakan
selected_columns = [
    'harga_awal_rp',
    'judul_layanan',
    'jumlah_layanan_ditawarkan',
    'jumlah_order',
    'jumlah_pelanggan',
    'jumlah_dipekerjakan_ulang',
    'skills',
    'kategori_keahlian',
    'rating_freelance'
]

df_selected = df[selected_columns]

print("Tampilan subset kolom yang dipilih:")
display(df_selected.head())
```

```python
# struktur data
print(df_selected.info())
```

```python
# cek missingvalue

print("Presentase Missing Value di Setiap Kolom:")
missing_count = df_selected.isnull().sum()

missing_percentage = (df_selected.isnull().sum() / len(df_selected)) * 100

missing_summary = pd.DataFrame({
    'Jumlah Missing': missing_count,
    'Persentase (%)': missing_percentage.round(2)
})
print(missing_summary)
```

```python
# cek duplikasi data
duplicate_all = df_selected.duplicated().sum()

print("Duplikasi seluruh baris:", duplicate_all)
```

```python
# Melihat beberapa baris yang duplikat (termasuk baris aslinya)
df_duplikat_contoh = df_selected[df_selected.duplicated(keep=False)]
display(df_duplikat_contoh.head(10))
```

```python
total_baris_awal = len(df_selected)
persentase_terbuang = (197 / total_baris_awal) * 100

print(f"Total baris awal: {total_baris_awal}")
print(f"Jumlah baris yang akan dihapus: 197")
print(f"Persentase data yang terbuang: {persentase_terbuang:.2f}%")
```
